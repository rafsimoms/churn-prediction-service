```python
from src.data.get_dataset import get_train_dataset
from src.models.validation import get_cv_splits
from src.models.preprocessing import preprocess
from src.models.evaluate import evaluate_model

from sklearn.dummy import DummyClassifier
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
import pandas as pd

df = get_train_dataset()
y = df['churned'].astype(int)
X = df.drop(columns=['churned', 'customer_id'])

scoring = ['f1', 'recall', 'roc_auc', 'precision', 'average_precision']

dummy_pipe = Pipeline([('prep', preprocess()), ('dummy', DummyClassifier(strategy='most_frequent'))])
logreg_pipe = Pipeline([('prep', preprocess()), ('logreg', LogisticRegression(max_iter=1000))])
tree_pipe = Pipeline([('prep', preprocess()), ('tree', DecisionTreeClassifier(max_depth=5, random_state=42, min_samples_leaf=20))])
cv_res_dummy = evaluate_model(pipeline=dummy_pipe, X=X, y=y, cv=get_cv_splits(), model_name='Dummy')
cv_res_logreg = evaluate_model(pipeline=logreg_pipe, X=X, y=y, cv=get_cv_splits(), model_name='Logreg')
cv_res_tree = evaluate_model(pipeline=tree_pipe, X=X, y=y, cv=get_cv_splits(), model_name='DecisionTree')
res = pd.concat([cv_res_tree, cv_res_logreg, cv_res_dummy])
res
```
